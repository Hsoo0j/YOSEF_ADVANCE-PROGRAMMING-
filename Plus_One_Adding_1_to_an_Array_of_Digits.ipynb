```python
# Title: Plus One - Adding 1 to an Array of Digits
# Student Name: [YOSEF CHRISTIAN]
# Date: [2026/03/14]

def plusOne(digits):
    """
    Adds one to a number represented as an array of digits

    Parameters:
    digits (list): List of integers representing the digits of a number

    Returns:
    list: The number incremented by 1, still as a list of digits
    """

    n = len(digits)

    # Process digits from right to left
    for i in range(n - 1, -1, -1):
        if digits[i] < 9:
            digits[i] += 1
            return digits
        else:  # digits[i] == 9
            digits[i] = 0
            # Continue loop to handle carry

    # If all digits were 9
    return [1] + digits

# Test Section
print("=" * 50)
print("TESTING THE PLUS ONE FUNCTION")
print("=" * 50)

# Test Case 1: Normal number
test1 = [1, 2, 3]
result1 = plusOne(test1.copy())  # Use .copy() to avoid modifying original
print(f"Test 1: {test1} + 1 = {result1} ✓")

# Test Case 2: Number ending with 9
test2 = [1, 2, 9]
result2 = plusOne(test2.copy())
print(f"Test 2: {test2} + 1 = {result2} ✓")

# Test Case 3: Multiple 9s in middle
test3 = [1, 9, 9]
result3 = plusOne(test3.copy())
print(f"Test 3: {test3} + 1 = {result3} ✓")

# Test Case 4: All 9s
test4 = [9, 9, 9]
result4 = plusOne(test4.copy())
print(f"Test 4: {test4} + 1 = {result4} ✓")

# Test Case 5: Single digit (not 9)
test5 = [5]
result5 = plusOne(test5.copy())
print(f"Test 5: {test5} + 1 = {result5} ✓")

# Test Case 6: From the problem statement
print("\n" + "=" * 50)
print("PROBLEM EXAMPLES")
print("=" * 50)
print(f"Example 1: [1,2,3] + 1 = {plusOne([1,2,3])} (Expected: [1,2,4])")
print(f"Example 2: [4,3,2,1] + 1 = {plusOne([4,3,2,1])} (Expected: [4,3,2,2])")
print(f"Example 3: [9] + 1 = {plusOne([9])} (Expected: [1,0])")

# Optional: Let user input their own number
print("\n" + "=" * 50)
print("TRY YOUR OWN NUMBER")
print("=" * 50)
user_input = input("Enter digits separated by spaces (e.g., 5 6 7): ")
digits_list = [int(x) for x in user_input.split()]
result = plusOne(digits_list.copy())
print(f"Your number: {digits_list}")
print(f"Plus one: {result}")
```
